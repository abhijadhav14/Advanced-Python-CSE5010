```python
class Student:
    #Class variable
    school = "Green Valley School"

    def __init__(self, name, age):
        self.name = name  
        self.age = age    

    def display(self):
        print(f"Name: {self.name}, Age: {self.age}")
        print(f"School: {Student.school}")

# access instance variables
s1 = Student("Alice", 20)
print(s1.name, s1.age)

# access class variable
print(Student.school)

#modifying instance variable
s1.name = "Charlie"
s1.age = 21
print(s1.name, s1.age)

# modifying class variable
Student.school = "Blue River School"
print(Student.school)
```

```python
# Encapsulation
#public

class Person:
    def __init__(self):
        self.name = "John"  

    def display_name(self):
        print(f"Name: {self.name}")

obj = Person()
obj.display_name()
print(obj.name)       
```

```python
#private

class Private1:
    def __init__(self):
        self.__salary = 50000    

    def salary(self):
        return self.__salary
    
obj = Private1()
print(obj.salary())
```

```python
#private

class Private1:
    def __init__(self):
        self.__salary = 50000
        self.__name = "Alice"    

    def salary(self):
        return self.__salary
    def name(self):
        return self.__name
    
    
obj = Private1()
print(obj.salary())
print(obj.name())
```

```python
class Student:

    def __init__(self, name, marks):
        self.__name = name
        self.__marks = marks

    # getter methods
    def get_name(self):
        return self.__name     

    def get_marks(self):
        return self.__marks
    
    # setter methods
    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks. Please enter between 0 and 100.")

    # change name
    def set_name(self, name):
        self.__name = name        

# Using the class
s = Student("David", 85)
print("Name:", s.get_name())
print("Marks:", s.get_marks())

s.set_marks(95)
print("Updated Marks:", s.get_marks())


s.set_name("Dwane")
print("Updated Name:", s.get_name())

s.set_marks(150) 
```

```python
# Create a class "bank" having data members customer and balance. 
# WPS to protect data members and perform following operation:
# Deposit, Withdraw, Show balance

class Bank:
    def __init__(self, customer, balance=0):
        self.__customer = customer
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: {amount}")
        else:
            print("Insufficient balance or invalid withdrawal amount.")

    def show_balance(self):
        print(f"Customer: {self.__customer}, Balance: {self.__balance}")


# Using the Bank class
b = Bank("Abhi", 1000)
b.show_balance()    
b.deposit(500)
b.show_balance()
b.withdraw(300)
b.show_balance()
b.withdraw(1500)
b.deposit(-200)
b.show_balance()
```
