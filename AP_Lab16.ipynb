```python
# OOPs in Python

class Person:
    def __init__(self, name, sex, profession):
        # data members (instance variables)
        self.name = name
        self.sex = sex
        self.profession = profession

    # Behavior (instance methods)    
    def show(self):
        print('Name:', self.name, 'Sex:', self.sex, 'Profession:', self.profession)

    # Behavior (instance methods)    
    def work(self):
        print(self.name, 'working as a', self.profession)

# create object of a class
jessa = Person('Jessa', 'Female', 'Software Engineer')
# call methods
jessa.show()
jessa.work()    
```

```python
# Create a class student having data members name rollno and age
# WAP to create student 2 objects and display both the students objects

class Student:
    def __init__(self, name, rollno, age):
        # data members (instance variables)
        self.name = name
        self.rollno = rollno
        self.age = age

    # Behavior (instance methods)    
    def show(self):
        print('Name:', self.name, 'Roll No:', self.rollno, 'Age:', self.age)

   
# create object of a class
s1 = Student('Ram', 101, 20)
s2 = Student('Jesus', 102, 21)

# call methods
s1.show()
s2.show()
```

```python
# Create a class Customer having data members name, custid, age and address
# WAP to display details of a customer

class Customer:
    def __init__(self, name, custid, age, address):
        # data members (instance variables)
        self.name = name
        self.custid = custid
        self.age = age
        self.address = address

    # Behavior (instance methods)    
    def show(self):
        print('Name:', self.name, 'CustID:', self.custid, 'Age:', self.age, 'Address:', self.address)

# create object of a class
c1 = Customer('Alice', 'C001', 30, '123 Main St')

# call methods
c1.show()
```

```python
# Class with user input
class Person:
    def __init__(self, name):
        self.name = name

    def show(self):
        print('Name:', self.name)

n = input('Enter name: ')
p = Person(n)
p.show()            
```

```python
# Students details from user

class Student:
    def __init__(self, name, rollno):
        # data members (instance variables)
        self.name = name
        self.rollno = rollno

    def show(self):
        print('Name:', self.name, 'Roll No:', self.rollno)

n = input('Enter name: ')
r = int(input('Enter roll no: '))
s = Student(n, r)
s.show()        
```
