```python
# python program to demonstrate
# Multilevel Inheritance

# Base class
class Grandfather:
    def __init__(self, grandfathername):
        self.grandfather_name = grandfathername

class Father(Grandfather):
    def __init__(self, grandfathername, father_name):
        super().__init__(grandfathername)
        self.father_name = father_name

        Grandfather.__init__(self, grandfathername)  

class Son(Father):
    def __init__(self, grandfathername, father_name, son_name):
        Father.__init__(self, grandfathername, father_name)
        self.son_name = son_name

    def print_names(self):
        print("Grandfather's name :", self.grandfather_name)
        print("Father's name :", self.father_name)
        print("Son's name :", self.son_name)

# Driver code
s1 = Son("John", "Michael", "David")
s1.print_names()
```

```python
# University -> Department -> Student
class University:
    def set_university(self):
        self.university_name = "Amity University"

class Department(University):
    def set_department(self):
        self.department_name = "Computer Science"

class Student(Department):
    def display(self):
        print("University Name :", self.university_name)
        print("Department Name :", self.department_name)
        print("Student Name : John") 

s = Student()
s.set_university()
s.set_department()
s.display()       
```

```python
# University -> Department -> Student
# user input
class University:
    def set_university(self):
        self.university_name = input("Enter University Name: ")

class Department(University):
    def set_department(self):
        self.department_name = input("Enter Department Name: ")
class Student(Department):
    def display(self):
        print("University Name :", self.university_name)
        print("Department Name :", self.department_name)
        print("Student Name :", self.student_name)

s = Student()
s.set_university()
s.set_department()
s.student_name = input("Enter Student Name: ")
s.display()     
```

```python
# An organization has security team, and each team has security analysts. Display cybersecurity-related details using multilevel inheritance.
class Organization:
    def set_organization(self):
        self.organization_name = input("Enter Organization Name: ")

class SecurityTeam(Organization):
    def set_team(self):
        self.team_name = input("Enter Security Team Name: ")

class SecurityAnalyst(SecurityTeam):
    def set_analyst(self):
        self.analyst_name = input("Enter Analyst Name: ")
        self.analyst_role = input("Enter Analyst Role: ")
        
    def display_details(self):
        print("Organization Name :", self.organization_name)
        print("Security Team Name :", self.team_name)
        print("Analyst Name :", self.analyst_name)
        print("Analyst Role :", self.analyst_role)

analyst = SecurityAnalyst()
analyst.set_organization()
analyst.set_team()
analyst.set_analyst()
analyst.display_details()
```

```python
# Lab Exam

# Suppose we have three classes Vehicle, Fourwheeler, and Car. The class Vehicle is the base class, the class Fourwheeler is derived from the Vehicle class, and the class Car is derived from the Fourwheeler class. Class Vehicle has method 'vehicle' that prints 'I am a vehicle', class Fourwheeler has method 'four_wheeler' that prints 'I have four wheels', and class Car has method 'car' that prints 'I am a car'. So, as this  is a multilevel inheritance; we can have access to all the other classes methods from the object of the class car. We invoke all the methods from a Car object and print the corresponding outputs of the methods. So, if we invoke the methods in this order, car(), four_wheeler(), vehicle(), the output will be:
# I am a car
# I have four wheels
# I am a vehicle
class Vehicle:
    def vehicle(self):
        print("I am a vehicle")

class Fourwheeler(Vehicle):
    def four_wheeler(self):
        print("I have four wheels")

class Car(Fourwheeler):
    def car(self):
        print("I am a car")
        
car_obj = Car()
car_obj.car()
car_obj.four_wheeler()
car_obj.vehicle()
```

```python
# Lab Exam

# 1.Define a class Vehicle with attributes brand and speed.
# Derive a class Car with fuelType.
# Display vehicle details.

class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

class Car(Vehicle):
    def __init__(self, brand, speed, fuelType):
        super().__init__(brand, speed)
        self.fuelType = fuelType

    def display_details(self):
        print("Vehicle Brand :", self.brand)
        print("Vehicle Speed :", self.speed)
        print("Car Fuel Type :", self.fuelType)
car = Car("Toyota", "180 km/h", "Petrol")
car.display_details()
```

```python
# 2. Create a class Account with accountNumber and balance.
# Derive SavingsAccount and add interestRate.
# Calculate interest.

class Account:
    def __init__(self, accountNumber, balance):
        self.accountNumber = input("Enter Account Number: ")
        self.balance = float(input("Enter Balance: "))

class SavingsAccount(Account):
    def __init__(self, accountNumber, balance, interestRate):
        super().__init__(accountNumber, balance)
        self.interestRate = interestRate

    def calculate_interest(self):
        interest = (self.balance * self.interestRate) / 100
        return interest
    
s = SavingsAccount("", 0, 5)
interest = s.calculate_interest()
print(f"Interest on Savings Account {s.accountNumber} with balance {s.balance}:", interest)
```
