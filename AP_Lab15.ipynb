```python
#MATPLOTLIB

import matplotlib.pyplot as plt
import numpy as np

xpoints = np.array([1, 8])
ypoints = np.array([3, 10])

plt.plot(xpoints, ypoints)
plt.title("Line Chart")

plt.gcf().patch.set_facecolor((1,1,1,1))
plt.show()
```

```python
# Plotting multiple points

import matplotlib.pyplot as plt
import numpy as np

x_points = np.array([1, 2, 6, 8])
y_points = np.array([3, 8, 1, 10])

plt.plot(x_points, y_points)
plt.title("Multiple Points")
plt.show()
```

```python
# Matplotlib Markers

import matplotlib.pyplot as plt
import numpy as np

y_points = np.array([3, 8, 1, 10])

plt.plot(y_points, marker = 'o')
plt.title("Markers")
plt.show()
```

```python
# Line Styles in Matplotlib

import matplotlib.pyplot as plt
import numpy as np

y_points = np.array([3, 8, 1, 10])

plt.plot(y_points, linestyle = 'dotted')
plt.title("Line Styles")
plt.show()
```

```python
# Matplotlib Labels and Title

import matplotlib.pyplot as plt
import numpy as np

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.plot(x, y)

plt.xlabel('Average Pulse')
plt.ylabel('Calorie Burnage')

plt.show()
```

```python
# Add Grid Lines

import matplotlib.pyplot as plt
import numpy as np

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.title('Sports Watch Data')
plt.xlabel('Average Pulse')
plt.ylabel('Calorie Burnage')

plt.plot(x, y)
plt.grid()
plt.show()
```

```python
# Bar Graph

import matplotlib.pyplot as plt
import numpy as np

xpoints = np.array([1,8])
ypoints = np.array([3,10])

plt.bar(xpoints, ypoints)
plt.title("Bar Graph")

plt.legend(["bar"])
plt.show()
```

```python
# Multiple Bar

import matplotlib.pyplot as plt
import numpy as np

xpoints = np.array([3, 1, 3, 12, 2, 4, 4])
ypoints = np.array([3, 2, 1, 4, 5, 6, 7])

plt.bar(xpoints, ypoints)
plt.title("Bar Chart")

plt.legend(["bar"])
plt.show()
```

```python
# Histogram

import matplotlib.pyplot as plt
import numpy as np  

x = [1, 2, 3, 4, 5, 6, 7, 4]

plt.hist(x, bins = [1,2,3,4,5,6,7])

plt.title("Histogram")
plt.legend(["bar"])
plt.show()
```

```python
# Scatter Plot

x = [3, 1, 3, 12, 2, 4, 4]
y = [3, 2, 1, 4, 5, 6, 7]
plt.scatter(x, y)

plt.legend("A")
plt.title("Scatter Plot")
plt.show()
```

```python
# Pie Chart

x = [1, 2, 3, 4]
e = [0.1, 0, 0, 0]

plt.pie(x, explode = e)

plt.title("Pie Chart")
plt.show()
```

```python
# Stem Plot

x = np.linspace(0.1, 2 * np.pi, 41)
y = np.exp(np.sin(x))

plt.stem(x, y)
plt.title("Stem Plot")
plt.show()
```

```python
# Error plot

x = [1,2,3,4,5,6,7]
y = [1,2,1,2,1,2,1]

y_error = 0.2

plt.plot(x, y)

plt.errorbar(x, y, yerr = y_error, fmt = 'o')
plt.title("Error Plot")
plt.show()
```
