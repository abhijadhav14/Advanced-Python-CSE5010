```python
# Create DataFrame from Multiple Series.

import pandas as pd

# Create two lists.
authors = ['Oscar', 'Max', 'Lewis', 'Nico']
articles = [300, 250, 400, 150]

# Creating two series by passing lists.
auth_series = pd.Series(authors)
article_series = pd.Series(articles)

# Creating a dictionary by passing series objects as values.
data_dict = {'authors': auth_series, 'articles': article_series}

# Creating DataFrame by passing dictionary.
result = pd.DataFrame(data_dict)

# Print the DataFrame.
print(result)
```

```python
# Create DataFrame from Multiple Series.

import pandas as pd

# Create two lists.
authors = ['Oscar', 'Max', 'Lewis', 'Nico']
articles = [300, 250, 400, 150]

# Creating two series by passing lists.
auth_series = pd.Series(authors)
article_series = pd.Series(articles)

# Creating a dictionary by passing series objects as values.
data_dict = {'authors': auth_series, 'articles': article_series}

# Creating DataFrame by passing dictionary.
result = pd.DataFrame(data_dict)

# Creating another list
age = [25, 30, 22, 28]

# Creating new column in DataFrame by providing a series created using list.
result['age'] = pd.Series(age)

# Print the DataFrame.
print(result)
```

```python
# A Teacher wants to calculate the average, Highest and Lowest marks of students in a class.

import pandas as pd

data = {
    'Students': pd.Series(['John', 'Emma', 'Sophia', 'Michael', 'Olivia']),
    'Marks': pd.Series([85, 92, 78, 90, 88])
}

df = pd.DataFrame(data)


print(df,"\n")

print("Average Marks:", df['Marks'].mean())
print("Highest Marks:", df['Marks'].max())
print("Lowest Marks:", df['Marks'].min())
```

```python
# An HR manager wants to find the employees earning above $60,000 and calculate the average salary.

import pandas as pd

data = {
    'Employees': pd.Series(['Alice', 'Bob', 'Charlie', 'David', 'Eva']),
    'Salaries': pd.Series([55000, 62000, 48000, 75000, 59000])
}

df = pd.DataFrame(data)

above_60k = df[df['Salaries'] > 60000]

print("Employees earning above $60,000:\n")
print(above_60k)
print("\nAverage Salary:", df['Salaries'].mean())
```
