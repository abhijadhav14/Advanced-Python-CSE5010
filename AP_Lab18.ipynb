```python
# Inheritance in python 

# Single Inheritance

# Base class
class Pr:
    def func1(self):
        print("This function is in Parent class")

    def func3(self):
        print("This function is also in Parent class")    

# Derived class
class Ch(Pr):
    def func2(self):
        print("This function is in Child class")

# Driver code   
ob = Ch()
ob.func1()   
ob.func3()
ob.func2()
```

```python
# parent class
class Person:
    def __init__(self, name, id):
        self.name = name
        self.id = id

    def display(self):
        print(self.name, self.id)
     

# child class
class Employee(Person):
    def print_emp(self):
        print("Employee class called")

emp1 = Employee("John", 101)
emp1.display()   
emp1.print_emp()
     
```

```python
class Employee:
    def __init__(self, name, emp_id):   
        self.name = name
        self.emp_id = emp_id

    def display(self):
        print(f"Name: {self.name}, ID: {self.emp_id}")


class Developer(Employee):
    #def __init__(self, name, emp_id, language):
    #    Employee.__init__(self, name, emp_id, domain="Software Development")
    #    self.language = language        

    def __init__(self, name, emp_id, domain):
        Employee.__init__(self, name, emp_id)
        self.domain = domain

    def show_domain(self):
        print(f"Domain: {self.domain}")

#usage
dev1 = Developer("Alice", 102, "Python")
dev2 = Developer("Bob", 103, "Java")
dev1.display()
dev1.show_domain()                
dev2.display()
dev2.show_domain()
```

```python
class Vehicle:
    def __init__(self, brand, max_speed):
        self.brand = brand
        self.max_speed = max_speed

    def show_info(self):
        print(f"Brand: {self.brand}, Max Speed: {self.max_speed} km/h")

class Car(Vehicle):
    def __init__(self, brand, max_speed, airbags):
        Vehicle.__init__(self, brand, max_speed)
        self.airbags = airbags

    def safety_info(self):
        print(f"Car has {self.airbags} airbags.")

# Usage
c = Car("Toyota", 180, 6)
c.show_info()
c.safety_info()
```

```python
# Multi-level Inheritance

class A:
    def showMe(self):
        print("I am class A")
class B(A):
    def showMe2(self):
        print("I am class B")
class C(B):
    def showMe3(self):
        print("I am class C")
obj = C()
obj.showMe()
obj.showMe2()
obj.showMe3()

b = B()
b.showMe()
b.showMe2()

a = A()
a.showMe()
```
