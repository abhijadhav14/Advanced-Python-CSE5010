```python
# WAPS to Create the following database with series and dataframe methods. Perform the following operations

# 1. Add a column color.

# 2. Display only the product column.

# 3. Display the product names having discount value > 10

import pandas as pd

data = [['Umbrella', 1250, 10],
        ['Mattress', 1450, 8],
        ['Badminton', 1550, 15],
        ['Shuttle', 400, 10]]

df = pd.DataFrame(data, columns=['product', 'Updated_Price', 'Discount'])
```

```python
# 1. Add a column color.
df['Color'] = ['Red', 'Blue', 'Green', 'Yellow']

print("DataFrame after adding Color column:")
print(df)
```

```python
# 2. Display only the product column.
product_column = df['product']
print("Product Column:")
print(product_column)
```

```python
# 3. Display the product names having discount value > 10
discounted_products = df[df['Discount'] > 10]['product']
print("\nProducts with Discount > 10:")
print(discounted_products)
```
